1. Polars

```python
import polars as pl
import pandas as pd
import numpy as np
import bottleneck as bn
```

```python
try:
    df_pl = pl.read_csv("train.csv")
    print("Датасет успешно прочитан с помощью Polars.")
except FileNotFoundError:
    print("Файл train.csv не найден. Убедитесь, что файл находится в текущей директории.")
    df_pl = None
```

```python
if df_pl is not None:
    print("\nОсновная информация о датасете (Polars):")
    print(df_pl.describe())
    print("\nТипы данных:")
    print(df_pl.dtypes)
    print("\nКоличество пропусков в каждом столбце:")
    for col in df_pl.columns:
        print(f"{col}: {df_pl[col].null_count()}")
```

```python
print("\nКоличество пассажиров каждого класса (Pclass) (Polars):")
print(df_pl.group_by("Pclass").agg(pl.count()).sort("Pclass"))
```

```python
print("\nКоличество выживших мужчин и женщин (Polars):")
print(df_pl.group_by(["Sex", "Survived"]).agg(pl.count()))
```

```python
print("\nПассажиры старше 44 лет (Polars):")
print(df_pl.filter(pl.col("Age") > 44))
```

Ускорение работы с pandas

```python
import pandas as pd
import numpy as np
import bottleneck as bn
```

```python
try:
    df = pd.read_csv("train.csv")
    print("Датасет успешно прочитан с помощью Pandas.")
except FileNotFoundError:
    print("Файл train.csv не найден. Убедитесь, что файл находится в текущей директории.")
    df = None
```

```python
if df is not None:
    print("\nСредний возраст пассажиров (Bottleneck):", bn.nanmean(df['Age']))
    print("Стандартное отклонение возраста пассажиров (Bottleneck):", bn.nanstd(df['Age']))
```

```python
df['Fare_new'] = np.where(df['Pclass'].isin([1, 2]), df['Fare'] * 1.3, df['Fare'] * 1.1)
print("\nПервые 5 строк с новым столбцом Fare_new:")
print(df.head())
```

Оптимизация типов pandas

```python
import pandas as pd
```

```python
try:
    df = pd.read_csv("Housing.csv")
    print("Датасет успешно прочитан с помощью Pandas.")
except FileNotFoundError:
    print("Файл Housing.csv не найден. Убедитесь, что файл находится в текущей директории.")
    df = None
```

```python
if df is not None:
  mem_before = df.memory_usage(deep=True).sum()
print(f"\nПотребление памяти до оптимизации: {mem_before / 1024**2:.2f} MB")
print("\nИнформация о типах данных до оптимизации:")
print(df.dtypes)
```

```python
print("\nДиапазоны и уникальные значения:")
for col in df.columns:
  if df[col].dtype in ['int64', 'float64']:
    print(f"{col}: min={df[col].min()}, max={df[col].max()}")
  else:
    print(f"{col}: unique values={df[col].unique()}")
```

```python
df['bedrooms'] = df['bedrooms'].astype('int8')
df['bathrooms'] = df['bathrooms'].astype('int8')
df['stories'] = df['stories'].astype('int8')
df['parking'] = df['parking'].astype('int8')
for col in ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']:
        df[col] = df[col].astype('category')
```

```python
df['price'] = df['price'].astype('int32')
df['area'] = df['area'].astype('int32')
```

```python
mem_after = df.memory_usage(deep=True).sum()
print(f"\nПотребление памяти после оптимизации: {mem_after / 1024**2:.2f} MB")
print(f"Экономия памяти: {((mem_before - mem_after) / mem_before) * 100:.2f}%")
print("\nИнформация о типах данных после оптимизации:")
print(df.dtypes)
```
